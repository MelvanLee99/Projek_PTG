```python
import pandas as pd
import random
from tabulate import tabulate
from termcolor import colored

def generate_weekly_schedule(m, n, p):
    """
    Generate a weekly learning schedule with side coloring.

    Parameters:
        m (int): Number of teachers.
        n (int): Number of classes.
        p (list of list): A 2D list where p[i][j] indicates the number of periods teacher X[i] teaches class Y[j].

    Returns:
        str: A formatted table showing the schedule.
    """

    # Colors for teachers (use distinct colors for differentiation)
    colors = ['red', 'green', 'yellow', 'blue', 'magenta', 'cyan', 'white']
    teacher_colors = {i: colors[i % len(colors)] for i in range(m)}

    # Create a DataFrame to represent the weekly schedule
    days = ["Senin", "Selasa", "Rabu", "Kamis", "Jumat"]
    schedule = pd.DataFrame(index=days, columns=[f"Y{j+1}" for j in range(n)])

    # Fill the schedule with teaching periods
    for i in range(m):
        for j in range(n):
            periods = p[i][j]
            if periods>0:
                assigned_days = random.sample(days, min(periods, len(days)))
                for day in assigned_days:
                    current_value = schedule.at[day, f"Y{j+1}"]
                    if pd.isna(current_value):
                        schedule.at[day, f"Y{j+1}"] = colored(f"X{i+1}", teacher_colors[i])
                    else:
                        schedule.at[day, f"Y{j+1}"] += f", {colored(f'X{i+1}', teacher_colors[i])}"

    # Generate the table with side coloring
    table = tabulate(schedule, headers='keys', tablefmt='grid', showindex=True)
    return table

# Example input
m = int(input("Enter the number of teachers: "))

# Input number of classes
n = int(input("Enter the number of classes: "))

# Initialize the periods matrix
p = []
print("Enter the number of periods for each teacher-class combination:")
for i in range(m):
    teacher_row = []
    for j in range(n):
        periods = int(input(f"X{i+1}, Y{j+1}: "))
        teacher_row.append(periods)
    p.append(teacher_row)
# Generate and print the schedule
schedule_table = generate_weekly_schedule(m, n, p)
print(schedule_table)
```
